```python
from data_manager import fetch_data
import pandas as pd

# Fetch 1-day OHLC data for Apple
symbol = ['^NSEI','HDFCBANK.NS','INFY.NS','ICICIBANK.NS','RELIANCE.NS','BHARTIARTL.NS']
data = pd.DataFrame()
for s in symbol:
    df = fetch_data(s, "2025-08-17", "2025-09-27", period="15m")
    cols = df.columns.get_level_values(0)
    cols = [c+'_'+s for c in cols]
    df.columns = cols
    data = pd.concat([data,df],axis=1)
data.index = data.index.tz_convert('Asia/Kolkata').tz_localize(None)    
print(data.head())
```

```python
for s in symbol:
    data['o_c_'+s] = data['Open_'+s]/data['Close_'+s] - 1
    data['h_l_'+s] = data['High_'+s]/data['Low_'+s] - 1
    
```

```python

import numpy as np
import statsmodels.api as sm

def rolling_ols(df, dep_col, window=30):
    # Identify dependent and independent variables
    y = df[dep_col]
    X = df.drop(columns=[dep_col])
    X = sm.add_constant(X)

    # Prepare containers
    results = {
        'r2': [],
        'dw': [],
        'predicted': [],
        'actual':[]
    }
    # Add placeholder for each beta and t-stat
    for col in X.columns:
        results[f'beta_{col}'] = []
        results[f'tstat_{col}'] = []

    # Rolling regression loop
    for i in range(len(df)):
        if i < window - 1:
            # Not enough data yet
            for key in results.keys():
                results[key].append(np.nan)
            continue
        
        y_window = y.iloc[i - window + 1:i + 1]
        X_window = X.iloc[i - window + 1:i + 1]

        model = sm.OLS(y_window, X_window).fit()
        
        # Append results
        for col in X.columns:
            results[f'beta_{col}'].append(model.params[col])
            results[f'tstat_{col}'].append(model.tvalues[col])

        results['r2'].append(model.rsquared)
        results['dw'].append(sm.stats.durbin_watson(model.resid))
        
        # Predicted value for the last observation in the window
        x_last = X.iloc[i:i+1]
        y_last = y_window[-1]
        results['predicted'].append(float(model.predict(x_last)))
        results['actual'].append(float(y_last))

    # Combine into DataFrame
    results_df = pd.DataFrame(results, index=df.index)
    return results_df

# Example usage:
# df = your dataframe containing o_c_^NSEI and the rest
reg_df=data[['o_c_'+s for s in symbol]]
reg_df = reg_df.replace([np.inf, -np.inf], np.nan).dropna()
results_df = rolling_ols(reg_df, dep_col='o_c_^NSEI', window=30)
```

```python
results_df = results_df.dropna()
```
